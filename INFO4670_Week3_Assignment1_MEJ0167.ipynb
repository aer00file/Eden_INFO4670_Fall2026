### Q1 · One quantitative variable
**How many hours per week do Northgate students work?** Describe what's typical, how spread out it is, and whether anyone stands apart.

```python
# Copy the closest cell from the Guided notebook, then fill in the blanks.
#
#   fig, ax = plt.subplots()
#   sns.__________(data=________, x="__________", color=NAVY, ax=ax)   # which chart? which column(s)?
#   ax.set_title("__________")
#   ax.set_xlabel("__________")
#   ax.set_ylabel("__________")
#   plt.show()

# --- your code below ---
fig, ax = plt.subplots()
sns.histplot(data=students, x="work_hours_per_week", color=NAVY, ax=ax)   # which chart? which column(s)?
ax.set_title("Work Hours Per Week for Northgate Students")
ax.set_xlabel("Work Hours")
ax.set_ylabel("Number of Students")
plt.show()
```

### Q2 · One categorical variable
**How does enrollment compare across majors?**
→ *sorted bar chart*. Column: `major` (from `students`).

```python
# Copy the closest cell from the Guided notebook, then fill in the blanks.
#
#   fig, ax = plt.subplots()
#   sns.__________(data=________, x="__________", color=NAVY, ax=ax)   # which chart? which column(s)?
#   ax.set_title("__________")
#   ax.set_xlabel("__________")
#   ax.set_ylabel("__________")
#   plt.show()

# --- your code below ---
order = students["major"].value_counts().index
fig, ax = plt.subplots()
sns.countplot(data=students, y="major", order=order, color=NAVY, ax=ax)   # which chart? which column(s)?
ax.set_title("Number of Northgate Students Enrolled in Each Major")
ax.set_xlabel("Major")
ax.set_ylabel("Number of Students")
plt.show()
```

**STAR story — Q2** *(write in your own words)*

- **Scope —** This chart shows the majors of all students
- **Track —** How many students are in each major?
- **Articulate —** The largest number of students are in cybersecurity, and the lowest number are in Information Science and Info Tech
- **Respond —** Nothing feels particularly noteworthy.

→ *box plot per major*. Columns: `final_gpa` and `major` (from `students`).

```python
# Copy the closest cell from the Guided notebook, then fill in the blanks.
#
#   fig, ax = plt.subplots()
#   sns.__________(data=________, x="__________", color=NAVY, ax=ax)   # which chart? which column(s)?
#   ax.set_title("__________")
#   ax.set_xlabel("__________")
#   ax.set_ylabel("__________")
#   plt.show()

# --- your code below ---

fig, ax = plt.subplots()
sns.boxplot(data=students, x="final_gpa", y="major",color=NAVY, ax=ax)   # which chart? which column(s)?
ax.set_title("Final GPA by Major")
ax.set_xlabel("Final GPA")
ax.set_ylabel("Major")
plt.show()
```

### Q4 · Two quantitative variables
**Is there a relationship between how far students commute and their final GPA?** Remember: *association is not causation.*
→ *scatterplot with a trend line*. Columns: `commute_miles` and `final_gpa` (from `students`).

```python
# Copy the closest cell from the Guided notebook, then fill in the blanks.
#
#   fig, ax = plt.subplots()
#   sns.__________(data=________, x="__________", color=NAVY, ax=ax)   # which chart? which column(s)?
#   ax.set_title("__________")
#   ax.set_xlabel("__________")
#   ax.set_ylabel("__________")
#   plt.show()

# --- your code below ---
fig, ax = plt.subplots()
sns.regplot(data=students, x="commute_miles", y="final_gpa", scatter_kws={"s": 12, "color": NAVY, "alpha": 0.4}, line_kws={"color": RUST},ax=ax)   # which chart? which column(s)?
ax.set_title("Final GPA vs Commute Miles")
ax.set_xlabel("Commute Miles")
ax.set_ylabel("Final GPA")
plt.show()
```

**STAR story — Q4** *(write in your own words)*

- **Scope —** This chart shows the final GPA and commute miles of all students.
- **Track —** Is there a relationship between final GPA and commute miles?
- **Articulate —** There is not a siginficant corelation between final GPA and commute miles.
- **Respond —** What would this chart look like if the 0.0 GPAs were removed?

### Q5 · Over time
**How does student engagement change across the semester?**
→ *line chart* of average weekly activity. Use the `weekly` table: `week` and `minutes_active` (you'll need to average by week first — see the Guided notebook).

```python
# Copy the closest cell from the Guided notebook, then fill in the blanks.
#
#   fig, ax = plt.subplots()
#   sns.__________(data=________, x="__________", color=NAVY, ax=ax)   # which chart? which column(s)?
#   ax.set_title("__________")
#   ax.set_xlabel("__________")
#   ax.set_ylabel("__________")
#   plt.show()

# --- your code below ---
weekly_avg = weekly.groupby("week")["minutes_active"].mean().reset_index()
fig, ax = plt.subplots()
sns.lineplot(data=weekly_avg, x="week", y="minutes_active", color=NAVY, marker="o", ax=ax)   # which chart? which column(s)?
ax.set_title("Average Engaged Minutes per Student, by week")
ax.set_xlabel("week")
ax.set_ylabel("Avg. engaged minutes")
plt.show()
```

**STAR story — Q5** *(write in your own words)*

- **Scope —** This chart shows the average engaged minutes of each student, categorized by week.
- **Track —** How does student engagement change over the semester?
- **Articulate —** Students generally spend over 200 minutes per week engaged, except for week 9, where they spend less than 100 minutes engaged.
- **Respond —** What is happening in week 9 to reduce enagement so much?
